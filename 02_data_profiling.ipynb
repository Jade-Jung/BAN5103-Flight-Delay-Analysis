```python
import pandas as pd
from pathlib import Path

raw_dir = Path("data/raw")

flights = pd.read_csv(raw_dir / "flights.csv")
weather = pd.read_csv(raw_dir / "weather.csv")
airlines = pd.read_csv(raw_dir / "airlines.csv")

print("Flights:", flights.shape)
print("Weather:", weather.shape)
print("Airlines:", airlines.shape)
```

```python
flight_missing = pd.DataFrame({
    "Missing Count": flights.isna().sum(),
    "Missing %": (flights.isna().mean() * 100).round(2)
})

display(
    flight_missing.sort_values(
        "Missing Count", ascending=False
    )
)
```

```python
print("Duplicate flight rows:", flights.duplicated().sum())
print("Duplicate weather rows:", weather.duplicated().sum())

print(
    "Duplicate weather join keys:",
    weather.duplicated(subset=["origin", "time_hour"]).sum()
)
```
